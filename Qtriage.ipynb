```python

from datetime import datetime, timedelta, timezone
from typing import Annotated

import jwt
from fastapi import Depends, FastAPI, HTTPException, status
from fastapi.security import OAuth2PasswordBearer, OAuth2PasswordRequestForm
from pydantic import BaseModel, EmailStr
from pwdlib import PasswordHash

app = FastAPI(title="Auth API")

SECRET_KEY = "change-this-to-a-long-random-secret-key"
ALGORITHM = "HS256"
ACCESS_TOKEN_EXPIRE_MINUTES = 30

password_hash = PasswordHash.recommended()
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="/auth/login")

users = {}


class RegisterRequest(BaseModel):
    name: str
    email: EmailStr
    password: str


class UserResponse(BaseModel):
    name: str
    email: EmailStr


def create_access_token(email: str):
    expires = datetime.now(timezone.utc) + timedelta(
        minutes=ACCESS_TOKEN_EXPIRE_MINUTES
    )

    payload = {
        "sub": email,
        "exp": expires,
    }

    return jwt.encode(payload, SECRET_KEY, algorithm=ALGORITHM)


@app.get("/")
def home():
    return {"message": "Auth API is running"}


@app.post("/auth/register", response_model=UserResponse)
def register(user: RegisterRequest):
    email = str(user.email).lower()

    if email in users:
        raise HTTPException(
            status_code=400,
            detail="Email already registered"
        )

    users[email] = {
        "name": user.name,
        "email": email,
        "password_hash": password_hash.hash(user.password),
    }

    return {
        "name": user.name,
        "email": email,
    }


@app.post("/auth/login")
def login(
    form_data: Annotated[OAuth2PasswordRequestForm, Depends()]
):
    email = form_data.username.lower()
    user = users.get(email)

    if not user or not password_hash.verify(
        form_data.password,
        user["password_hash"]
    ):
        raise HTTPException(
            status_code=status.HTTP_401_UNAUTHORIZED,
            detail="Incorrect email or password",
            headers={"WWW-Authenticate": "Bearer"},
        )

    token = create_access_token(email)

    return {
        "access_token": token,
        "token_type": "bearer",
    }


@app.get("/auth/me", response_model=UserResponse)
def get_me(token: Annotated[str, Depends(oauth2_scheme)]):
    try:
        payload = jwt.decode(
            token,
            SECRET_KEY,
            algorithms=[ALGORITHM]
        )
        email = payload.get("sub")

        if not email or email not in users:
            raise HTTPException(
                status_code=401,
                detail="Invalid token"
            )

        user = users[email]

        return {
            "name": user["name"],
            "email": user["email"],
        }

    except jwt.PyJWTError:
        raise HTTPException(
            status_code=401,
            detail="Invalid or expired token"
        )
```

```python

!pip install fastapi uvicorn python-jose passlib pydantic[email] pyngrok -q

from fastapi import FastAPI, HTTPException, Depends, status
from fastapi.security import OAuth2PasswordBearer
from pydantic import BaseModel, EmailStr
from passlib.context import CryptContext
from jose import jwt, JWTError
from datetime import datetime, timedelta
from typing import Optional

app = FastAPI()

SECRET_KEY = "your-secret-key-here"
ALGORITHM = "HS256"

pwd_context = CryptContext(schemes=["bcrypt"], deprecated="auto")
users = []

class UserRegister(BaseModel):
    name: str
    email: EmailStr
    password: str

class UserLogin(BaseModel):
    email: EmailStr
    password: str

class Token(BaseModel):
    access_token: str
    token_type: str

class UserResponse(BaseModel):
    email: str
    name: str

def hash_password(password: str) -> str:
    return pwd_context.hash(password)

def verify_password(plain: str, hashed: str) -> bool:
    return pwd_context.verify(plain, hashed)

def create_access_token(data: dict, expires_delta: timedelta = timedelta(hours=1)):
    to_encode = data.copy()
    expire = datetime.utcnow() + expires_delta
    to_encode.update({"exp": expire})
    return jwt.encode(to_encode, SECRET_KEY, algorithm=ALGORITHM)

@app.post("/register")
def register(user: UserRegister):
    for u in users:
        if u["email"] == user.email:
            raise HTTPException(status_code=400, detail="Email already exists")

    users.append({
        "name": user.name,
        "email": user.email,
        "password": hash_password(user.password)
    })
    return {"message": "Registration successful", "email": user.email}

@app.post("/login", response_model=Token)
def login(user: UserLogin):
    for u in users:
        if u["email"] == user.email:
            if not verify_password(user.password, u["password"]):
                raise HTTPException(status_code=401, detail="Invalid credentials")
            token = create_access_token({"sub": u["email"]})
            return {"access_token": token, "token_type": "bearer"}
    raise HTTPException(status_code=401, detail="Invalid credentials")

@app.get("/me", response_model=UserResponse)
def get_me(token: str = Depends(OAuth2PasswordBearer(tokenUrl="/login"))):
    try:
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        email: str = payload.get("sub")
        if not email:
            raise HTTPException(status_code=401, detail="Invalid token")
    except JWTError:
        raise HTTPException(status_code=401, detail="Invalid token")

    for u in users:
        if u["email"] == email:
            return {"email": u["email"], "name": u["name"]}
    raise HTTPException(status_code=404, detail="User not found")
```

```python

# ==============================
# QTRIAGE BACKEND - RECOVERED
# ==============================

!pip -q install fastapi uvicorn pydantic

import sqlite3
import threading
import time
import uvicorn

from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from typing import Optional
from datetime import datetime

# ------------------------------
# DATABASE
# ------------------------------

DB = "qtriage.db"

conn = sqlite3.connect(DB, check_same_thread=False)
cursor = conn.cursor()

cursor.executescript("""
CREATE TABLE IF NOT EXISTS patients (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    phone TEXT
);

CREATE TABLE IF NOT EXISTS doctors (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT
);

CREATE TABLE IF NOT EXISTS appointments (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    patient_name TEXT,
    doctor_name TEXT,
    appointment_time TEXT,
    status TEXT
);

CREATE TABLE IF NOT EXISTS queue (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    patient_name TEXT,
    doctor_name TEXT,
    token_number INTEGER
);

CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    email TEXT UNIQUE,
    password TEXT,
    role TEXT
);

CREATE TABLE IF NOT EXISTS issues (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    patient_name TEXT,
    description TEXT,
    status TEXT
);
""")
conn.commit()

# ------------------------------
# FASTAPI
# ------------------------------

app = FastAPI(
    title="Qtriage Backend",
    description="Qtriage Hospital Management Backend API",
    version="1.0"
)

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

# ------------------------------
# MODELS
# ------------------------------

class Patient(BaseModel):
    name: str
    age: Optional[int] = None
    phone: str

class Doctor(BaseModel):
    name: str
    department: Optional[str] = None

class Appointment(BaseModel):
    patient_name: str
    doctor_name: str
    appointment_time: str
    status: str = "scheduled"

class QueueEntry(BaseModel):
    patient_name: str
    doctor_name: str
    token_number: int

class UserRegister(BaseModel):
    name: str
    email: str
    password: str

class UserLogin(BaseModel):
    email: str
    password: str

class RoleUser(BaseModel):
    email: str
    role: str

class Issue(BaseModel):
    patient_name: str
    description: str
    status: str = "reported"

class PredictionRequest(BaseModel):
    symptoms: str
    age: int

# ------------------------------
# BASIC
# ------------------------------

@app.get("/")
def home():
    return {"message": "Qtriage Backend is Running!"}

@app.get("/health")
def health():
    return {"status": "healthy"}

@app.get("/api/status")
def api_status():
    return {
        "project": "Qtriage",
        "backend": "FastAPI",
        "status": "running"
    }

# ------------------------------
# PATIENTS
# ------------------------------

@app.post("/patients")
def create_patient(p: Patient):
    cursor.execute(
        "INSERT INTO patients(name,age,phone) VALUES(?,?,?)",
        (p.name, p.age, p.phone)
    )
    conn.commit()
    return {"id": cursor.lastrowid, **p.model_dump()}

@app.get("/patients")
def get_patients():
    rows = cursor.execute("SELECT * FROM patients").fetchall()
    return [
        {"id":r[0], "name":r[1], "age":r[2], "phone":r[3]}
        for r in rows
    ]

@app.get("/patients/{patient_id}")
def get_patient(patient_id: int):
    r = cursor.execute(
        "SELECT * FROM patients WHERE id=?", (patient_id,)
    ).fetchone()

    if not r:
        raise HTTPException(404, "Patient not found")

    return {"id":r[0], "name":r[1], "age":r[2], "phone":r[3]}

@app.put("/patients/{patient_id}")
def update_patient(patient_id: int, p: Patient):
    cursor.execute(
        "UPDATE patients SET name=?,age=?,phone=? WHERE id=?",
        (p.name,p.age,p.phone,patient_id)
    )
    conn.commit()

    if cursor.rowcount == 0:
        raise HTTPException(404, "Patient not found")

    return {"message":"Patient updated successfully"}

@app.delete("/patients/{patient_id}")
def delete_patient(patient_id: int):
    cursor.execute("DELETE FROM patients WHERE id=?", (patient_id,))
    conn.commit()

    if cursor.rowcount == 0:
        raise HTTPException(404, "Patient not found")

    return {"message":"Patient deleted successfully"}

# ------------------------------
# DOCTORS
# ------------------------------

@app.post("/doctors")
def create_doctor(d: Doctor):
    cursor.execute(
        "INSERT INTO doctors(name,department) VALUES(?,?)",
        (d.name,d.department)
    )
    conn.commit()
    return {"id":cursor.lastrowid, **d.model_dump()}

@app.get("/doctors")
def get_doctors():
    rows = cursor.execute("SELECT * FROM doctors").fetchall()
    return [
        {"id":r[0],"name":r[1],"department":r[2]}
        for r in rows
    ]

@app.get("/doctors/{doctor_id}")
def get_doctor(doctor_id:int):
    r=cursor.execute(
        "SELECT * FROM doctors WHERE id=?", (doctor_id,)
    ).fetchone()

    if not r:
        raise HTTPException(404,"Doctor not found")

    return {"id":r[0],"name":r[1],"department":r[2]}

@app.put("/doctors/{doctor_id}")
def update_doctor(doctor_id:int,d:Doctor):
    cursor.execute(
        "UPDATE doctors SET name=?,department=? WHERE id=?",
        (d.name,d.department,doctor_id)
    )
    conn.commit()

    if cursor.rowcount==0:
        raise HTTPException(404,"Doctor not found")

    return {"message":"Doctor updated successfully"}

@app.delete("/doctors/{doctor_id}")
def delete_doctor(doctor_id:int):
    cursor.execute("DELETE FROM doctors WHERE id=?", (doctor_id,))
    conn.commit()

    if cursor.rowcount==0:
        raise HTTPException(404,"Doctor not found")

    return {"message":"Doctor deleted successfully"}

# ------------------------------
# APPOINTMENTS
# ------------------------------

@app.post("/appointments")
def create_appointment(a:Appointment):
    cursor.execute(
        """INSERT INTO appointments
        (patient_name,doctor_name,appointment_time,status)
        VALUES(?,?,?,?)""",
        (a.patient_name,a.doctor_name,a.appointment_time,a.status)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**a.model_dump()}

@app.get("/appointments")
def get_appointments():
    rows=cursor.execute("SELECT * FROM appointments").fetchall()
    return [
        {
            "id":r[0],
            "patient_name":r[1],
            "doctor_name":r[2],
            "appointment_time":r[3],
            "status":r[4]
        } for r in rows
    ]

@app.put("/appointments/{appointment_id}")
def update_appointment(appointment_id:int,a:Appointment):
    cursor.execute(
        """UPDATE appointments
        SET patient_name=?,doctor_name=?,appointment_time=?,status=?
        WHERE id=?""",
        (a.patient_name,a.doctor_name,a.appointment_time,a.status,appointment_id)
    )
    conn.commit()

    if cursor.rowcount==0:
        raise HTTPException(404,"Appointment not found")

    return {"message":"Appointment updated successfully"}

@app.delete("/appointments/{appointment_id}")
def delete_appointment(appointment_id:int):
    cursor.execute(
        "DELETE FROM appointments WHERE id=?", (appointment_id,)
    )
    conn.commit()

    if cursor.rowcount==0:
        raise HTTPException(404,"Appointment not found")

    return {"message":"Appointment deleted successfully"}

# ------------------------------
# QUEUE
# ------------------------------

@app.post("/queue")
def add_queue(q:QueueEntry):
    cursor.execute(
        """INSERT INTO queue
        (patient_name,doctor_name,token_number)
        VALUES(?,?,?)""",
        (q.patient_name,q.doctor_name,q.token_number)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**q.model_dump()}

@app.get("/queue")
def get_queue():
    rows=cursor.execute("SELECT * FROM queue").fetchall()
    return [
        {
            "id":r[0],
            "patient_name":r[1],
            "doctor_name":r[2],
            "token_number":r[3]
        } for r in rows
    ]

# ------------------------------
# AUTH
# ------------------------------

@app.post("/register")
def register(u:UserRegister):
    try:
        cursor.execute(
            "INSERT INTO users(name,email,password,role) VALUES(?,?,?,?)",
            (u.name,u.email,u.password,"patient")
        )
        conn.commit()
        return {"message":"Registration successful","role":"patient"}
    except sqlite3.IntegrityError:
        raise HTTPException(400,"Email already registered")

@app.post("/login")
def login(u:UserLogin):
    r=cursor.execute(
        "SELECT id,name,email,role FROM users WHERE email=? AND password=?",
        (u.email,u.password)
    ).fetchone()

    if not r:
        raise HTTPException(401,"Invalid email or password")

    return {
        "message":"Login successful",
        "user_id":r[0],
        "name":r[1],
        "email":r[2],
        "role":r[3]
    }

@app.post("/role")
def set_role(u:RoleUser):
    cursor.execute(
        "UPDATE users SET role=? WHERE email=?",
        (u.role,u.email)
    )
    conn.commit()

    return {"message":"Role updated","email":u.email,"role":u.role}

# ------------------------------
# ISSUES
# ------------------------------

@app.post("/issues")
def create_issue(i:Issue):
    cursor.execute(
        "INSERT INTO issues(patient_name,description,status) VALUES(?,?,?)",
        (i.patient_name,i.description,i.status)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**i.model_dump()}

@app.get("/issues")
def get_issues():
    rows=cursor.execute("SELECT * FROM issues").fetchall()
    return [
        {
            "id":r[0],
            "patient_name":r[1],
            "description":r[2],
            "status":r[3]
        } for r in rows
    ]

@app.put("/issues/{issue_id}/status")
def update_issue(issue_id:int,status:str):
    cursor.execute(
        "UPDATE issues SET status=? WHERE id=?",
        (status,issue_id)
    )
    conn.commit()

    if cursor.rowcount==0:
        raise HTTPException(404,"Issue not found")

    return {"message":"Issue status updated"}

# ------------------------------
# AI PREDICTION
# ------------------------------

@app.post("/ai/predict")
def ai_predict(p:PredictionRequest):
    symptoms=p.symptoms.lower()

    if any(x in symptoms for x in ["chest pain","breathing","severe"]):
        priority="high"
    elif any(x in symptoms for x in ["fever","headache","pain"]):
        priority="medium"
    else:
        priority="low"

    return {
        "priority":priority,
        "note":"Demo prediction for Qtriage"
    }

# ------------------------------
# DASHBOARD
# ------------------------------

@app.get("/dashboard/summary")
def dashboard_summary():
    patients=cursor.execute(
        "SELECT COUNT(*) FROM patients"
    ).fetchone()[0]

    doctors=cursor.execute(
        "SELECT COUNT(*) FROM doctors"
    ).fetchone()[0]

    appointments=cursor.execute(
        "SELECT COUNT(*) FROM appointments"
    ).fetchone()[0]

    queue_size=cursor.execute(
        "SELECT COUNT(*) FROM queue"
    ).fetchone()[0]

    return {
        "total_patients":patients,
        "total_doctors":doctors,
        "total_appointments":appointments,
        "queue_size":queue_size
    }

# ------------------------------
# START SERVER
# ------------------------------

def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000)

thread = threading.Thread(target=run_server, daemon=True)
thread.start()

time.sleep(3)

print("================================")
print("✅ Qtriage Backend Started")
print("✅ FastAPI running on port 8000")
print("✅ Swagger: /docs")
print("================================")
```

```python

%%writefile "/content/drive/MyDrive/CBSE - Senior School/Qtriage/Qtriage_backend.py"

import sqlite3
import threading
import time
import uvicorn

from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from typing import Optional

DB = "qtriage.db"

conn = sqlite3.connect(DB, check_same_thread=False)
cursor = conn.cursor()

cursor.executescript("""
CREATE TABLE IF NOT EXISTS patients (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    phone TEXT
);

CREATE TABLE IF NOT EXISTS doctors (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT
);

CREATE TABLE IF NOT EXISTS appointments (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    patient_name TEXT,
    doctor_name TEXT,
    appointment_time TEXT,
    status TEXT
);

CREATE TABLE IF NOT EXISTS queue (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    patient_name TEXT,
    doctor_name TEXT,
    token_number INTEGER
);

CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    email TEXT UNIQUE,
    password TEXT,
    role TEXT
);

CREATE TABLE IF NOT EXISTS issues (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    patient_name TEXT,
    description TEXT,
    status TEXT
);
""")
conn.commit()

app = FastAPI(
    title="Qtriage Backend",
    description="Qtriage Hospital Management Backend API",
    version="1.0"
)

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"]
)

class Patient(BaseModel):
    name: str
    age: Optional[int] = None
    phone: str

class Doctor(BaseModel):
    name: str
    department: Optional[str] = None

class Appointment(BaseModel):
    patient_name: str
    doctor_name: str
    appointment_time: str
    status: str = "scheduled"

class QueueEntry(BaseModel):
    patient_name: str
    doctor_name: str
    token_number: int

class UserRegister(BaseModel):
    name: str
    email: str
    password: str

class UserLogin(BaseModel):
    email: str
    password: str

class RoleUser(BaseModel):
    email: str
    role: str

class Issue(BaseModel):
    patient_name: str
    description: str
    status: str = "reported"

class PredictionRequest(BaseModel):
    symptoms: str
    age: int

@app.get("/")
def home():
    return {"message": "Qtriage Backend is Running!"}

@app.get("/health")
def health():
    return {"status": "healthy"}

@app.get("/api/status")
def api_status():
    return {
        "project": "Qtriage",
        "backend": "FastAPI",
        "status": "running"
    }

@app.post("/patients")
def create_patient(p: Patient):
    cursor.execute(
        "INSERT INTO patients(name,age,phone) VALUES(?,?,?)",
        (p.name,p.age,p.phone)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**p.model_dump()}

@app.get("/patients")
def get_patients():
    rows=cursor.execute("SELECT * FROM patients").fetchall()
    return [
        {"id":r[0],"name":r[1],"age":r[2],"phone":r[3]}
        for r in rows
    ]

@app.get("/patients/{patient_id}")
def get_patient(patient_id:int):
    r=cursor.execute(
        "SELECT * FROM patients WHERE id=?",(patient_id,)
    ).fetchone()
    if not r:
        raise HTTPException(404,"Patient not found")
    return {"id":r[0],"name":r[1],"age":r[2],"phone":r[3]}

@app.put("/patients/{patient_id}")
def update_patient(patient_id:int,p:Patient):
    cursor.execute(
        "UPDATE patients SET name=?,age=?,phone=? WHERE id=?",
        (p.name,p.age,p.phone,patient_id)
    )
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Patient not found")
    return {"message":"Patient updated successfully"}

@app.delete("/patients/{patient_id}")
def delete_patient(patient_id:int):
    cursor.execute("DELETE FROM patients WHERE id=?",(patient_id,))
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Patient not found")
    return {"message":"Patient deleted successfully"}

@app.post("/doctors")
def create_doctor(d:Doctor):
    cursor.execute(
        "INSERT INTO doctors(name,department) VALUES(?,?)",
        (d.name,d.department)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**d.model_dump()}

@app.get("/doctors")
def get_doctors():
    rows=cursor.execute("SELECT * FROM doctors").fetchall()
    return [
        {"id":r[0],"name":r[1],"department":r[2]}
        for r in rows
    ]

@app.get("/doctors/{doctor_id}")
def get_doctor(doctor_id:int):
    r=cursor.execute(
        "SELECT * FROM doctors WHERE id=?",(doctor_id,)
    ).fetchone()
    if not r:
        raise HTTPException(404,"Doctor not found")
    return {"id":r[0],"name":r[1],"department":r[2]}

@app.put("/doctors/{doctor_id}")
def update_doctor(doctor_id:int,d:Doctor):
    cursor.execute(
        "UPDATE doctors SET name=?,department=? WHERE id=?",
        (d.name,d.department,doctor_id)
    )
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Doctor not found")
    return {"message":"Doctor updated successfully"}

@app.delete("/doctors/{doctor_id}")
def delete_doctor(doctor_id:int):
    cursor.execute("DELETE FROM doctors WHERE id=?",(doctor_id,))
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Doctor not found")
    return {"message":"Doctor deleted successfully"}

@app.post("/appointments")
def create_appointment(a:Appointment):
    cursor.execute(
        """INSERT INTO appointments
        (patient_name,doctor_name,appointment_time,status)
        VALUES(?,?,?,?)""",
        (a.patient_name,a.doctor_name,a.appointment_time,a.status)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**a.model_dump()}

@app.get("/appointments")
def get_appointments():
    rows=cursor.execute("SELECT * FROM appointments").fetchall()
    return [
        {
            "id":r[0],
            "patient_name":r[1],
            "doctor_name":r[2],
            "appointment_time":r[3],
            "status":r[4]
        } for r in rows
    ]

@app.put("/appointments/{appointment_id}")
def update_appointment(appointment_id:int,a:Appointment):
    cursor.execute(
        """UPDATE appointments
        SET patient_name=?,doctor_name=?,appointment_time=?,status=?
        WHERE id=?""",
        (a.patient_name,a.doctor_name,a.appointment_time,a.status,appointment_id)
    )
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Appointment not found")
    return {"message":"Appointment updated successfully"}

@app.delete("/appointments/{appointment_id}")
def delete_appointment(appointment_id:int):
    cursor.execute(
        "DELETE FROM appointments WHERE id=?",(appointment_id,)
    )
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Appointment not found")
    return {"message":"Appointment deleted successfully"}

@app.post("/queue")
def add_queue(q:QueueEntry):
    cursor.execute(
        """INSERT INTO queue
        (patient_name,doctor_name,token_number)
        VALUES(?,?,?)""",
        (q.patient_name,q.doctor_name,q.token_number)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**q.model_dump()}

@app.get("/queue")
def get_queue():
    rows=cursor.execute("SELECT * FROM queue").fetchall()
    return [
        {
            "id":r[0],
            "patient_name":r[1],
            "doctor_name":r[2],
            "token_number":r[3]
        } for r in rows
    ]

@app.post("/register")
def register(u:UserRegister):
    try:
        cursor.execute(
            "INSERT INTO users(name,email,password,role) VALUES(?,?,?,?)",
            (u.name,u.email,u.password,"patient")
        )
        conn.commit()
        return {"message":"Registration successful","role":"patient"}
    except sqlite3.IntegrityError:
        raise HTTPException(400,"Email already registered")

@app.post("/login")
def login(u:UserLogin):
    r=cursor.execute(
        "SELECT id,name,email,role FROM users WHERE email=? AND password=?",
        (u.email,u.password)
    ).fetchone()
    if not r:
        raise HTTPException(401,"Invalid email or password")
    return {
        "message":"Login successful",
        "user_id":r[0],
        "name":r[1],
        "email":r[2],
        "role":r[3]
    }

@app.post("/role")
def set_role(u:RoleUser):
    cursor.execute(
        "UPDATE users SET role=? WHERE email=?",
        (u.role,u.email)
    )
    conn.commit()
    return {"message":"Role updated","email":u.email,"role":u.role}

@app.post("/issues")
def create_issue(i:Issue):
    cursor.execute(
        "INSERT INTO issues(patient_name,description,status) VALUES(?,?,?)",
        (i.patient_name,i.description,i.status)
    )
    conn.commit()
    return {"id":cursor.lastrowid,**i.model_dump()}

@app.get("/issues")
def get_issues():
    rows=cursor.execute("SELECT * FROM issues").fetchall()
    return [
        {
            "id":r[0],
            "patient_name":r[1],
            "description":r[2],
            "status":r[3]
        } for r in rows
    ]

@app.put("/issues/{issue_id}/status")
def update_issue(issue_id:int,status:str):
    cursor.execute(
        "UPDATE issues SET status=? WHERE id=?",
        (status,issue_id)
    )
    conn.commit()
    if cursor.rowcount==0:
        raise HTTPException(404,"Issue not found")
    return {"message":"Issue status updated"}

@app.post("/ai/predict")
def ai_predict(p:PredictionRequest):
    symptoms=p.symptoms.lower()

    if any(x in symptoms for x in ["chest pain","breathing","severe"]):
        priority="high"
    elif any(x in symptoms for x in ["fever","headache","pain"]):
        priority="medium"
    else:
        priority="low"

    return {
        "priority":priority,
        "note":"Demo prediction for Qtriage"
    }

@app.get("/dashboard/summary")
def dashboard_summary():
    patients=cursor.execute("SELECT COUNT(*) FROM patients").fetchone()[0]
    doctors=cursor.execute("SELECT COUNT(*) FROM doctors").fetchone()[0]
    appointments=cursor.execute("SELECT COUNT(*) FROM appointments").fetchone()[0]
    queue_size=cursor.execute("SELECT COUNT(*) FROM queue").fetchone()[0]

    return {
        "total_patients":patients,
        "total_doctors":doctors,
        "total_appointments":appointments,
        "queue_size":queue_size
    }

print("✅ Qtriage backend code loaded")
```
